## Mori Model

Transient graded stimulus at t=0s and a similar reversed stimulus at t=200s

Replication of Fig. 2c from Mori et al. (2008)

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from matplotlib.animation import PillowWriter
from pde_utils import double_derivative

# define parameters
Da = 0.1 # A diffusion rate
Db = 10 # B diffusion rate
delta = 1 # basal rate of inactivation
gamma = 1 # GTPase autocatalytic activation rate
K = 1 # saturation parameter
k0 = 0.067 # basal GEF conversion rate
size = 100 # number of cells
dt = 0.01  # time step
t = 1000.0 # run time
frame_int = 5 # animation frame interval
a0 =  0.2683312
b0 = 2.0

# set initial concentrations
A = np.full(size, a0)
B = np.full(size, b0)

# create forward/backward gradients
x1 = np.linspace(10, 0, size)
x2 = np.linspace(0, 10, size)

# stimulus amplitude
S = 0.02

# define double gradient stimulus function
def s(t):
    if 0 <= t < 20:
        return S * x1
    elif 20 <= t < 25:
        return S * (1 - (t - 20) / 5)  *  x1
    elif 200 <= t < 220:
        return S * x2
    elif 220 <= t < 225:
        return S * (1 - (t - 220) / 5) * x2
    else:
        return 0

# define reaction kinetics function
def f(a, b):
    return (k0 + gamma*a**2/(K**2 + a**2))*b - delta*a

# set up figure
fig, ax = plt.subplots(figsize=(8, 4))
frames = []

# simulate the PDE with finite difference method
for i in range(int(t/dt)):

    kgrad = s(i*dt)

    # update concentrations
    A = A + dt * (Da * double_derivative(A) + f(A, B) + kgrad)
    B = B + dt * (Db * double_derivative(B) - f(A, B) - kgrad)

    # apply Neumann boundary conditions
    for Z in (A, B):
        Z[0] = Z[1]
        Z[-1] = Z[-2]

    # capture a frame every frame_int seconds
    if i % int(frame_int/dt) == 0:
        ALine, = ax.plot(A, color='blue')
        BLine, = ax.plot(B, color="red")
        KLine, = ax.plot(kgrad, color="green")
        frames.append([ALine, BLine, KLine])

animation.ArtistAnimation(fig, frames, interval=100, blit=True).save('animations/mori_double_graded_stimulus.gif', writer=PillowWriter(fps=10))
```
