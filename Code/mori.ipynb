## Mori Model

(2.1) from Hughes et al. 2024

$\frac{\partial u}{\partial t}=(b+\gamma\frac{u^n}{1+u^n})v-Iu+D\frac{\partial^2u}{\partial x^2}$

$\frac{\partial v}{\partial t}=-(b+\gamma\frac{u^n}{1+u^n})v+Iu+\frac{\partial^2v}{\partial x^2}$

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from matplotlib.animation import PillowWriter
from pde_utils import double_derivative

# define parameters
I = 1 # basal inactivation rate
Da = 0.1 # A diffusion rate
Db = 10 # B diffusion rate
beta = 4 # basal rate of activation
n = 2 # Hill Coefficient
gamma = 1 # GTPase autocatalytic activation rate
K = 1 # saturation parameter
k0 = 0.067 # basal GEF conversion rate

size = 100 # number of cells
dt = 0.01  # time step
t = 2000.0 # run time
frame_int = 25 # animation frame interval

# set initial concentrations
A = 0.1+0.9*np.random.rand(size) # active GTPase
B = np.full(size, 0.1) # inactive GTPase

# define reaction kinetics function
def f(a,b):
    return (k0 + gamma*a**2/(K^2+a^2))*b-beta*a

# set up figure
fig, ax = plt.subplots(figsize=(8, 4))
frames = []

# simulate the PDE with finite difference method
for i in range(int(t/dt)):

    # update concentrations
    U = U + dt * (f(A,B) + Da*double_derivative(U))
    V = V + dt * (-f(A,B) + Db*double_derivative(V))

    # capture a frame every frame_int seconds
    if i % int(frame_int/dt) == 0:
        Uline, = ax.plot(A, color='blue')
        Vline, = ax.plot(B, color='red')
        frames.append([Uline, Vline])

# save animation
animation.ArtistAnimation(fig, frames, interval=50, blit=True).save('animations/mori.gif', writer=PillowWriter(fps=10))
```
